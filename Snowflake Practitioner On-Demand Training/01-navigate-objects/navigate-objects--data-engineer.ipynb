## Session Variables
The cell below calls `get_active_session()` — a Snowpark function that gives Python the **same Snowflake session your SQL cells use**. From it we read your username and build a few variable *names* (`my_db`, `my_wh`, `my_role`) that follow the course naming convention.

> ⚠️ At this point these are just Python strings — building them doesn't activate anything. The **next cell (SQL)** is what actually runs `USE ROLE` / `USE WAREHOUSE` to set your session context, so the role/warehouse picker at the top-right won't update until that SQL runs.

```
from snowflake.snowpark.context import get_active_session
session = get_active_session()
user = session.get_current_user().strip('"')
my_db = user + '_DB'
my_wh = user + '_WH'
my_role = user + '_RL'
my_analyst_role = user + '_ANALYST_RL'
print('Context ready. Signed in as ' + user)
print('Your workspace -> db: ' + my_db + ' | wh: ' + my_wh + ' | role: ' + my_role)
```

### Your angle — Data Engineer.
Build a tiny end-to-end pipeline in your own database: a **landing** table, a **transform** into a curated table, and a **view** as the clean read layer.

> 💡 **Why this matters for engineers:** This landing → transform → serve pattern is the skeleton of every production pipeline you'll build. Getting it right at small scale means fewer rewrites at large scale.

> 👉 **Step 1 — land.** Create a staging schema and an empty target shaped like `ORDERS`, ready to load into later.

```
CREATE OR REPLACE TABLE {{my_db}}.STAGING.ORDERS_LANDING LIKE PLATPLUS_DB.ROASTERY_SALES.ORDERS;
DESCRIBE TABLE {{my_db}}.STAGING.ORDERS_LANDING;
```

> 🔎 **Review the output.** You've just *provisioned a landing zone* — an empty table shaped exactly like `ORDERS`, ready to receive raw data. Standing up a landing table before any data arrives is a standard first move in a pipeline, and it's a good example of this lab's theme: creating objects that follow common patterns.
>
> 📌 **One thing to notice:** the transform in the next step reads from the shared `ORDERS` table, not from your empty `ORDERS_LANDING`. *Filling* a landing table is its own discipline — file formats, stages, `COPY` — so this course gives it dedicated attention rather than rushing it here. For now, focus on the shape of the pipeline you're building.

> 👉 **Step 2 — transform.** Curate the shared raw orders into a daily-sales table you own (`CREATE TABLE AS SELECT`).

```
CREATE OR REPLACE TABLE {{my_db}}.STAGING.DAILY_SALES AS
SELECT store_id,
       DATE(order_ts)             AS order_date,
       COUNT(*)                   AS order_count,
       ROUND(SUM(total_amount),2) AS total_sales
FROM PLATPLUS_DB.ROASTERY_SALES.ORDERS
GROUP BY store_id, DATE(order_ts);
```

```
SELECT * FROM {{my_db}}.STAGING.DAILY_SALES ORDER BY order_date DESC, total_sales DESC LIMIT 10;
```

> 🔎 **Review the output.** One row per store per day — you turned raw orders into a curated table with a single CTAS.

> 👉 **Step 3 — serve.** A [view](https://docs.snowflake.com/en/sql-reference/sql/create-view) is a saved query: the clean read layer analysts use without touching your tables.

```
CREATE OR REPLACE VIEW {{my_db}}.STAGING.V_SALES_BY_DAY AS
SELECT order_date,
       SUM(order_count)          AS orders,
       ROUND(SUM(total_sales),2) AS total_sales
FROM {{my_db}}.STAGING.DAILY_SALES
GROUP BY order_date;
```

```
SELECT * FROM {{my_db}}.STAGING.V_SALES_BY_DAY ORDER BY order_date DESC LIMIT 10;
```

> 🔎 **Review the output.** Landing → curated table → view: that's the shape of nearly every pipeline you'll build. You've now created an object at each stage — the exact kind of pattern-based object creation this lab is about.

## Check Your Work
Run the next cell to **check your work**. It queries your own database to confirm you created the objects and rows this lab asked for, and prints a ✅ / ⛔ result for each — no need to read the code.

```
from snowflake.snowpark.context import get_active_session
session = get_active_session()
checks = [
    ('Your SANDBOX.MY_TASTING_NOTES has at least 2 tasting rows', f'SELECT COUNT(*) FROM {my_db}.SANDBOX.MY_TASTING_NOTES', 2),
]
passed = 0
for label, sql, need in checks:
    try:
        n = session.sql(sql).collect()[0][0]
        ok = n >= need
    except Exception:
        n, ok = None, False
    passed += 1 if ok else 0
    icon = '✅' if ok else '⛔'
    detail = f'(found {n})' if n is not None else '(not found)'
    print(f'{icon}  {label}  {detail}')
print('-' * 52)
if passed == len(checks):
    print(f'🎉 ALL {len(checks)} CHECKS PASSED — lab complete! Nice work.')
else:
    print(f'⚠️  {passed}/{len(checks)} passed — re-run the cells above for any ⛔ item, then run this cell again.')
```
